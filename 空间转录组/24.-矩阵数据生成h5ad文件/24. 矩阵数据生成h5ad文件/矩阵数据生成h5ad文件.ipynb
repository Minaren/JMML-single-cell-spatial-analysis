# 百创S系列空间转录组数据生成h5ad文件

## 概述
在单细胞与空间组学研究领域，h5ad文件作为Anndata的核心数据格式，已成为Scanpy等主流分析工具的通用接口。然而，百创智造的S系列空间转录组芯片（S1000/S2000/S3000/S4000/S5000）凭借其亚细胞级分辨率和“三片合一”技术，但Scanpy默认无法直接解析其空间坐标与图片信息。针对这一痛点，我们开发了定制化脚本工具，无缝打通百创数据与Scanpy生态，让科研人员无需重复造轮子即可开展高阶分析。  
Scanpy作为单细胞分析的金标准工具，其h5ad文件格式虽支持存储基因表达矩阵、细胞注释和空间坐标，但百创S系列数据的空间信息格式与常规Visium等平台存在结构性差异：  

空间坐标文件命名规则不同
组织切片图像元数据存储位置特殊
细胞/Spot ID对齐逻辑需定制化处理
若手动修改，不仅耗时费力，还容易引入人为错误，影响数据可重复性。

## 脚本参数

```
python bmk_spatial_h5ad_v2.py --help
Usage: bmk_spatial_h5ad_v2.py [options]

Generates Spatial h5ad

Options:
  -h, --help            show this help message and exit
  -i INPUT, --input=INPUT, --in=INPUT
                        BSTViewer_project/subdata/L13_heAuto/ [default: .]
  --png=PNG             BSTViewer_project/he_roi_small.png
  --library=LIBRARY     library id
  --type=TYPE           The options are 'S1000', 'S2000A', 'S2000B', 'S3000',
                        'cell_split'
  -o OUTPUT, --output=OUTPUT, --out=OUTPUT
                        where to output the gene count matrix [default: .]

-i： 指定矩阵文件目录

--png： png格式图片路径

--library：指定library id

--type： 数据类型，必须指定，可选参数有'S1000', 'S2000A', 'S2000B', 'S3000', 'cell_split'

-o： 指定输出目录
```
